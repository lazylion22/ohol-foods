```python
from bs4 import BeautifulSoup
import json

# load your HTML file
with open("foods.html", "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

data = []

for obj in soup.select(".objectList .object"):
    name = obj.select_one("h3").get_text(strip=True)
    
    link = obj.select_one("a")["href"]
    
    value = obj.select_one(".badgeValue").get_text(strip=True)
    
    img = obj.select_one("img[id^=image]")["src"]

    data.append({
        "name": name,
        "link": link,
        "value": int(value.split()[0]),
        "image": img
    })

# save to JSON
with open("foods.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("Done.")
```
